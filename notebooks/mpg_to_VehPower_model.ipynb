```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# matching mean, min, and max of VehPower and mpg of cars extracted
x_data = np.array([18, 32, 113])
y_data = np.array([113, 6.5, 4])

# points already look like exponential
def exponential_model(x, a, b, c):
    return a * np.exp(-b * x) + c

# fit the exponential
popt, pcov = curve_fit(exponential_model, x_data, y_data, p0=[100, 0.1, 4])

a_opt, b_opt, c_opt = popt
print(f"Optimized Parameters:")
print(f"a (Amplitude) = {a_opt:.4f}")
print(f"b (Decay Rate) = {b_opt:.4f}")
print(f"c (Asymptote) = {c_opt:.4f}")

x_plot = np.linspace(10, 120, 100)
y_plot = exponential_model(x_plot, *popt)

plt.scatter(x_data, y_data, color='red', label='Data Points')
plt.plot(x_plot, y_plot, label=f'Fit: $y={a_opt:.0f}e^{{-{b_opt:.2f}x}} + {c_opt:.1f}$')
plt.legend()
plt.grid(True)
plt.show()
```
